```python
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/cat_image.jpg")

cv2_imshow(image)
```

```python
print("SHAPE:", image.shape)
```

5472 - height; 3078 - width; 3 - color channels, BGR

```python
(b, g, r) = image[100, 50]
print("Blue:", b, "Green:", g, "Red:", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized = cv2.resize(gray, (300, 300))
cv2_imshow(resized)
```

```python
h, w = image.shape[:2]
cropped = image[int(h * 0.25):int(h * 0.55), int(w * 0.15):int(w * 0.85)]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(gray, 1)
cv2_imshow(flipped)
```

```python
cv2.imwrite("flipped_cat.jpg", flipped)
```
