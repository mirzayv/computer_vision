```python
# Food-101 ni kagglehub orqali yuklab, kerakli klasslarni Drive'ga nusxalash
import kagglehub, os, shutil

path = kagglehub.dataset_download("kmader/food41")
print("Path to dataset files:", path)

images_dir = None
for root, dirs, files in os.walk(path):
    if 'hot_dog' in dirs:
        images_dir = root
        break
print("images_dir:", images_dir)

TARGET_ROOT = '/content/drive/MyDrive/food101_dataset'
for cls in ['hamburger', 'hot_dog', 'pizza', 'cheesecake', 'chocolate_cake']:
    shutil.copytree(os.path.join(images_dir, cls),
                    os.path.join(TARGET_ROOT, cls),
                    dirs_exist_ok=True)
    print(cls, "->", len(os.listdir(os.path.join(TARGET_ROOT, cls))), "images")
```

```python
import os, requests
from io import BytesIO
from PIL import Image
from ddgs import DDGS

def download_images(query, target_dir, limit=100):
    os.makedirs(target_dir, exist_ok=True)
    results = DDGS().images(query,
                            max_results=limit * 2,
                            safesearch="on",
                            type_image="photo",
                            backend="duckduckgo")
    count = 0
    for r in results:
        if count >= limit:
            break
        try:
            resp = requests.get(r["image"], timeout=10,
                                headers={"User-Agent": "Mozilla/5.0"})
            img = Image.open(BytesIO(resp.content)).convert("RGB")
            if min(img.size) < 100:
                continue
            count += 1
            img.save(os.path.join(target_dir, f"Image_{count}.jpg"), "JPEG")
            print(f"[{count}] {r['image'][:90]}")
        except Exception as e:
            print("skip:", type(e).__name__)
    print(f"Done. {count} images -> {target_dir}")

download_images("uzbek pilaf plov food", '/content/drive/MyDrive/food101_dataset/pilaf')
download_images("kebab shashlik food", '/content/drive/MyDrive/food101_dataset/kebab')
```
